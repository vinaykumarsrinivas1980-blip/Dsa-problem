```python
#Write a program to reverse a string without using any inbuilt functions.

def reverse_string(s):
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str

print(reverse_string("hello"))

```

```python
#Write a program to replace one string with another.
#Example: "Hello World" → replace "World" with "Universe". only logic no code

def replace_string(s, old, new):
    result = ""
    i = 0

    while i < len(s):
        # Check if substring matches the word to replace
        if s[i:i+len(old)] == old:
            result += new
            i += len(old)   # Skip the length of old word
        else:
            result += s[i]
            i += 1

    return result


s = "Hello World"
print(replace_string(s, "World", "Universe"))
```

```python
# Find the 2nd largest element in array

def seclarge(num):
    n = len(num)

    if n < 2:
        return None   # Not enough elements

    largest = float('-inf')
    second_largest = float('-inf')

    for value in num:

        if value > largest:
            second_largest = largest
            largest = value

        elif value < largest and value > second_largest:
            second_largest = value

    if second_largest == float('-inf'):
        return None   # No second largest (all elements same)

    return second_largest


num = [55, 32, -97, -99, 3, 67]
print(seclarge(num))

```
