```python
#Write a program to reverse a string without using any inbuilt functions.

def reverse_string(s):
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str

print(reverse_string("hello"))

```

```python
#Write a program to replace one string with another.
#Example: "Hello World" → replace "World" with "Universe". only logic no code

def replace_string(s, old, new):
    result = ""
    i = 0

    while i < len(s):
        # Check if substring matches the word to replace
        if s[i:i+len(old)] == old:
            result += new
            i += len(old)   # Skip the length of old word
        else:
            result += s[i]
            i += 1

    return result


s = "Hello World"
print(replace_string(s, "World", "Universe"))
```

```python
# Find the 2nd largest element in array

def seclarge(num):
    n = len(num)

    if n < 2:
        return None   # Not enough elements

    largest = float('-inf')
    second_largest = float('-inf')

    for value in num:

        if value > largest:
            second_largest = largest
            largest = value

        elif value < largest and value > second_largest:
            second_largest = value

    if second_largest == float('-inf'):
        return None   # No second largest (all elements same)

    return second_largest


num = [55, 32, -97, -99, 3, 67]
print(seclarge(num))

```

```python
# right shift by k places

num = [1,2,3,4,5,6,7,8]

def rotat(num):
    k = 3
    n = len(num)

    k = k % n   # important optimization

    while k > 0:
        temp = num[n-1]

        for i in range(n-2, -1, -1):
            num[i+1] = num[i]

        num[0] = temp
        k -= 1   # decrease after one full rotation

    print(num)

rotat(num)
```

```python
#MOVE 0 TO LEFT brute force

num=[1,0,3,4,5,6,0,9,0,4,0,4]

def mov(num):
  n=len(num)
  temp=[]
  for i in range(0,n):
    if num[i]!=0:
      temp.append(num[i])
  nl=len(temp)

  for i in range(0,nl):
    num[i]=temp[i]
  for i in range(nl,n):
    num[i]=0

  print(num)
mov(num)
```

```python
#linear search

num=[1,2,3,1,1,3,4,4]

def linear(num,target):
  n=len(num)

  for i in range(0,n):
    if num[i]==target:
      return 1

  return 0
target=4
linear(num,target)
```
